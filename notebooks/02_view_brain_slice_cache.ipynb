# Brain Slice MEA Cache Viewer

Load a cache directory, inspect lightweight summaries, and write large interactive Plotly views to HTML files.

```python
from pathlib import Path
import pandas as pd

from acute_slice_mea.cache import load_band_power, load_cache_manifest, load_npz_dict
from acute_slice_mea.plots import write_band_power_html, write_spectrum_summary_html, write_trace_preview_html

cache_dir = Path("../data/processed/brain_slice_well004_cache")
manifest = load_cache_manifest(cache_dir)
manifest["summary"]
```

```python
electrodes = pd.read_csv(manifest["files"]["electrodes"])
band_power = load_band_power(manifest["files"]["band_power"])
spectrum = load_npz_dict(manifest["files"]["spectrum"])
trace_preview = load_npz_dict(manifest["files"]["trace_preview"])

display(electrodes.head())
display(band_power.head())
band_power.groupby("band")["power_db"].describe()
```

```python
figures_dir = cache_dir / "figures"
paths = {
    "band_power": write_band_power_html(band_power, figures_dir / "viewer_lfp_band_power.html"),
    "spectrum": write_spectrum_summary_html(spectrum, figures_dir / "viewer_lfp_spectrum_summary.html"),
    "trace_preview": write_trace_preview_html(trace_preview, figures_dir / "viewer_trace_preview_raw_lfp_spike.html"),
}
for name, path in paths.items():
    print(f"{name}: {path}")
```

```python
# Small inline summary plot: mean band power by band.
ax = band_power.groupby("band", observed=True)["power_db"].mean().plot(kind="bar", figsize=(7, 3), title="Mean LFP band power")
ax.set_ylabel("Power (dB)")
```
