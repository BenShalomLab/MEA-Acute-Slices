# Brain Slice MEA Cache Computation

Configure one raw Maxwell data file and one well, then run the heavy analysis pipeline. Outputs are written to the selected cache directory.

```python
from pathlib import Path
import json
import sys

project_root = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
src_path = project_root / "src"
if src_path.exists() and str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))
loaded_package = sys.modules.get("acute_slice_mea")
loaded_package_file = getattr(loaded_package, "__file__", None)
if loaded_package_file and src_path.resolve() not in Path(loaded_package_file).resolve().parents:
    for module_name in [name for name in sys.modules if name == "acute_slice_mea" or name.startswith("acute_slice_mea.")]:
        del sys.modules[module_name]

from acute_slice_mea import AnalysisConfig, run_analysis

data_path = "../data/external/data.raw.h5"
well_id = "well004"
output_dir = "../data/processed/brain_slice_well004_cache"

config = AnalysisConfig(
    data_path=data_path,
    well_id=well_id,
    output_dir=output_dir,
    lfp_low_hz=0.5,
    lfp_high_hz=300,
    spike_low_hz=300,
    spike_high_hz=3000,
    lfp_window_sec=10,
    lfp_step_sec=5,
    welch_segment_sec=2,
    spectrum_duration_sec=10,
    spectrum_max_freq_hz=150,
    preview_start_sec=0,
    preview_duration_sec=10,
    preview_max_points=20000,
    preview_max_electrodes=16,
    dashboard_max_points_per_electrode=20000,
    export_dashboard_data=True,
    spikeinterface_chunk_duration="60s",
    preview_electrode_ids=None,
    n_jobs=-2,
    channel_chunk_size=None,
    progress=True,
    verbose=False,
)

config
```

```python
manifest = run_analysis(config)
print(json.dumps(manifest["summary"], indent=2))
print("\nGenerated files:")
for key, value in manifest["files"].items():
    print(f"{key}: {value}")
```
