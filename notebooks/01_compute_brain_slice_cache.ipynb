# Brain Slice MEA Cache Computation

Configure one raw Maxwell data file and one well, then run the heavy analysis pipeline. Outputs are written to the selected cache directory.

```python
from pathlib import Path
import json

from acute_slice_mea import AnalysisConfig, run_analysis

data_path = "../data/external/data.raw.h5"
well_id = "well004"
output_dir = "../data/processed/brain_slice_well004_cache"

config = AnalysisConfig(
    data_path=data_path,
    well_id=well_id,
    output_dir=output_dir,
    lfp_low_hz=0.5,
    lfp_high_hz=300,
    spike_low_hz=300,
    spike_high_hz=3000,
    lfp_window_sec=10,
    lfp_step_sec=5,
    welch_segment_sec=2,
    spectrum_duration_sec=10,
    spectrum_max_freq_hz=150,
    preview_start_sec=0,
    preview_duration_sec=10,
    preview_max_points=20000,
    preview_max_electrodes=16,
    dashboard_max_points_per_electrode=20000,
    export_dashboard_data=True,
    spikeinterface_chunk_duration="60s",
    preview_electrode_ids=None,
)

config
```

```python
manifest = run_analysis(config)
print(json.dumps(manifest["summary"], indent=2))
print("\nGenerated files:")
for key, value in manifest["files"].items():
    print(f"{key}: {value}")
```
